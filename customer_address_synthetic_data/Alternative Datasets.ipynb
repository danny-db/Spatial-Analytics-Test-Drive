Other Customer Address Datasets to explore:

* [Vicmap Address](https://www.land.vic.gov.au/maps-and-spatial/spatial-data/vicmap-catalogue/vicmap-address), but this is specific to addresses located in Victoria, Australia.

* [Geoscape G-NAF](https://data.gov.au/data/dataset/geocoded-national-address-file-g-naf), a geocoded address database for Australian businesses and governments.

* [Overture](https://docs.overturemaps.org/getting-data/data-mirrors/databricks/), available for free via the Databricks Marketplace. These datasets are published and maintained by CARTO.
