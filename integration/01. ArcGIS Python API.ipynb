```python
#import the GIS class
from arcgis import GIS

# gis = GIS(url="abc", username="abc", password="abc")
gis = GIS()
#print("Loggined in as: " + gis.properties.user.username)
```

https://www.arcgis.com/home/item.html?id=942d52abe9104273b2a3a5a33762672f

```python
state_bndry_item = gis.content.get("942d52abe9104273b2a3a5a33762672f")
state_bndry_item
```

```python
# obtain the feature layer
slayer = state_bndry_item.layers[0]

# get a Spatially Enabled DataFrame
import pandas as pd
sdf = pd.DataFrame.spatial.from_layer(slayer)

# get the shape
display(sdf.shape)

sdf.head()
```

```python
df = spark.createDataFrame(sdf)
```

```python
display(df)
```

```python
(
  df.write
  .mode("overwrite")
  .option("overwriteSchema", "true")
  .saveAsTable("danny_catalog.demo_schema.tree_points_bronze")
)
```

```python
%sql
SELECT DISTINCT(commodity) FROM danny_catalog.demo_schema.tree_points_bronze
```

```python
%sql
CREATE OR REPLACE TABLE danny_catalog.demo_schema.tree_points_silver AS
SELECT *, 
  ST_AsText(ST_GeomFromWKB(SHAPE)) AS SHAPE_WKT,
  ST_X(ST_GeomFromWKB(SHAPE)) AS longitude,
  ST_Y(ST_GeomFromWKB(SHAPE)) AS latitude
FROM danny_catalog.demo_schema.tree_points_bronze
```

```python
%sql
SELECT * FROM danny_catalog.demo_schema.tree_points_silver
```
