# Load Bronze Layer

Datasets in scope includes:
- Transmission data, provided by the Digital Atlas of Australia:
    * [Electricity Transmission Lines, Digital Atlas of Australia](https://digital.atlas.gov.au/datasets/70f23e91102a4d6899a776d093fa08ef_2)
    * [Transmission Substations, Digital Atlas of Australia](https://digital.atlas.gov.au/datasets/d5eae2d7c9e54581a5f19d7e95b9883b_0)
    * [Major Power Stations, Digital Atlas of Australia](https://digital.atlas.gov.au/datasets/3d0f2d1b8aec4e1a8870b03ce11d4405_1)
- Digital Boundary e.g. State and Territory polygon. This is used to filter the transmission network to VIC state. Here, we're loading in the digital boundary files provided by the Australian Bureau of Statistics:
    * [Australian Statistical Geography Standard (ASGS) Edition 3, Australian Bureau of Statistics](https://www.abs.gov.au/statistics/standards/australian-statistical-geography-standard-asgs-edition-3/jul2021-jun2026/access-and-downloads/digital-boundary-files#downloads-for-gda94-digital-boundary-files)

```python
# Load Transmission data by using geopandas library to process polygon geometry data, convert to a Pandas and then Spark DataFrame, and finally writing it to a Delta table

import geopandas as gpd
import pandas as pd

volume_loc = f"/Volumes/{CATALOG}/{ELEC_SCHEMA}/{ELEC_VOLUME}"
datasets = {
  # Key: Table Name, Value = "Location in Volume"
  "major_power_station": "Major_Power_Stations/Major_Power_Stations.shp",
  "transmission_line": "Electricity_Transmission_Lines/Electricity_Transmission_Lines.shp",
  "transmission_substation": "Transmission_Substations/Transmission_Substations.shp"
}

for table_name, file_loc in datasets.items():
  gdf = gpd.read_file(f"{volume_loc}/{file_loc}")
  gdf['geometry'] = gdf['geometry'].to_wkt()
  pdf = pd.DataFrame(gdf)
  sdf = spark.createDataFrame(pdf)

  sdf.write.mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{CATALOG}.{ELEC_SCHEMA}.bronze_{table_name}")
```
