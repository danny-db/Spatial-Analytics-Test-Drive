# 05 · Grant the app read access
The Databricks App runs as its own service principal. This step looks that principal up by app name and grants it
read access to the schema (tables + the `heat_exposure` function). It's safe to re-run.

```python
%run ./00_config
```

```python
dbutils.widgets.text("app_name", "climate-exposure-explorer", "App name")
APP_NAME = dbutils.widgets.get("app_name")

from databricks.sdk import WorkspaceClient
sp = WorkspaceClient().apps.get(APP_NAME).service_principal_client_id
print(f"App '{APP_NAME}' runs as service principal {sp}")

for stmt in [f"GRANT USE CATALOG ON CATALOG {CATALOG} TO `{sp}`",
             f"GRANT USE SCHEMA, SELECT, EXECUTE ON SCHEMA {FQ} TO `{sp}`"]:
    spark.sql(stmt)
    print("✓", stmt)
display(spark.sql(f"SHOW GRANTS `{sp}` ON SCHEMA {FQ}"))
```
